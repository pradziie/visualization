```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100,50)

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)

ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
```

```python
font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14}

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1, loc = 'left')
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y,
marker= '*',
markersize = 12, # NEW ADDITIONS
color = '#7425b9',
linestyle = '--',
linewidth = 2,
markeredgecolor = '#fa9359', # NEW ADDITIONS
markerfacecolor = '#000000' ) # NEW ADDITIONS

ax.grid(axis = 'y')
ax.grid(axis = 'y', color = "blue", linewidth = 2, linestyle = '-.')
```

```python
# Create a dataframe
value1=np.random.uniform(size=20)
value2=value1+np.random.uniform(size=20)/4
df = pd.DataFrame({'group':list(map(chr, range(65, 85))), 'lowest performer':value1 , 'highest performer':value2 })
 
# Reorder it following the values of the first value:
ordered_df = df.sort_values(by='lowest performer')
my_range=range(1,len(df.index)+1)
 
# The horizontal plot is made using the hline function
fig, ax = plt.subplots(figsize=(7, 5))
ax.hlines(y=my_range, xmin=ordered_df['lowest performer'], xmax=ordered_df['highest performer'], color='grey', alpha=0.4, zorder=1)
ax.scatter(ordered_df['lowest performer'], my_range, color='#DC4633', alpha=1, label='lowest performer')
ax.scatter(ordered_df['highest performer'], my_range, color='#1E3765', alpha=1 , label='highest performer')
ax.legend()

# Add title and axis names
ax.set_yticks(my_range)
ax.set_yticklabels(ordered_df['group'])
ax.set_title("Comparison of the lowest and the highest performers in the teams", loc='left')
ax.set_xlabel('Score')
ax.set_ylabel('Team')

# color those spines!
ax.spines['top'].set_color('#919191')
ax.spines['right'].set_color('#919191')
ax.spines['left'].set_color("#919191")
ax.spines['bottom'].set_color('#919191')

# coloring those minor ticks and major ticks
ax.xaxis.set_tick_params(which='major', size=7, width=2, color='#919191')
ax.xaxis.set_tick_params(which='minor', size=4, color='#919191')

# Add grid only to x axis
ax.grid(axis = 'x')

# Show the graph
plt.show()
```
