```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100, 50)
y2 = np.random.randint(0, 100, 50)

```

```python
# plotting
fig, ax = plt.subplots(figsize=(5, 3)) # figure size in inches
ax.plot(x,y1, label = 'Person 1') # plot with label for legend
ax.plot(x,y2, label = 'Person 2') 
ax.legend(loc = 'lower right', # legend location WITHIN THE LEGEND BOX
          frameon = True, # legend frame on/off. Frame means the box around the legend
          fontsize = 12, # legend font size
          ncol = 2, # number of columns in legend
          shadow = True) # legend shadow on/off
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc="upper left",
          bbox_to_anchor=(0.5, 0.5)) # anchoring at 50% of x and 50% of y of the axes
fig.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
np.random.seed(613)
```

```python
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
plt.style.use('default')
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(7, 3))
# make smaller font size in callouts/ticks
for label in (ax1.get_xticklabels() + ax1.get_yticklabels() + ax2.get_xticklabels() + ax2.get_yticklabels()):
    label.set_fontsize(10)
# change colors of bars

ax1.scatter(x1, y1)
ax2.bar(x2, y2)
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
['ax2', 'ax3']],
figsize=(7, 4))
someaxes["ax1"].scatter(x1,y1, color='#DC4633')
someaxes["ax2"].bar(x2,y2, color='#1E3765')
someaxes["ax3"].plot(x1,y1, color='#6D9B8A')
plt.show()
```

```python
# subplot mosaic - abomination style!
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax4']],
                                    figsize=(7, 4),
                                    layout = 'constrained') # CONSTRAINING THIS... failed to contain the greatness of this layout
someaxes["ax1"].scatter(x1,y1, color='#DC4633')
someaxes["ax2"].bar(x2,y2, color='#1E3765')
someaxes["ax3"].plot(x1,y1, color='#007FA3')
someaxes["ax4"].plot(x2,y2, color='#F1C500')


someaxes["ax1"].set_xlabel('A Big Label for Ax1', fontsize=18)
someaxes["ax2"].set_xlabel('Another label for Ax2', fontsize=18)
someaxes["ax3"].set_xlabel('BIGBADABOOM', fontsize=32)
someaxes["ax4"].set_xlabel('LALALALALALALA', fontsize=120)
```

```python
# first make our sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])
# define our figure and axes (just one this time)
fig, ax = plt.subplots(figsize=(7, 3))
# now call both bar and plot elements to the same axes (ax)
ax.bar(x, y1,
color = "#1E3765")
ax.plot(x, y2,
color = "#DC4633")
```

```python
y2_sd = np.std(y2) # this is our standard deviation
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "#1E3765")

# now adding an error bar
ax.errorbar(x, #our x values
y2, #our y values
yerr = y2_sd,
fmt = "none", # fmt is format of the main plot element, "none" means no marker
ecolor = '#DC4633',
elinewidth= 4,
capsize = 6,
capthick= 4,
errorevery=2) # show error bar every 2 data points
```

```python
from PIL import Image # to open images
import requests # to get images from URLs
from io import BytesIO # to store images in memory

# Changed to use my image ^_^
response = requests.get("https://www.gunjap.net/site/wp-content/uploads/2014/02/228.jpg") 
image_file = BytesIO(response.content)
image = Image.open(image_file)
```

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "#DC4633") # always UofT red-ish!

ax_image = fig.add_axes([0.1, # x coordinate (ON FIGURE, NOT AXES)
                        0.11, # y coordinate (ON FIGURE, NOT AXES)
                        0.15, # image width
                        0.35]) # image height) 
ax_image.imshow(image)
ax_image.axis('off') # turn off axis for image
```
